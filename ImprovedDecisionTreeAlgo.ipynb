 # DATASET
 

```python
data = [
    {"Age": "<=30",   "Income": "high",   "Student": "no",  "Credit_rating": "fair",      "buys_computer": "no"},
    {"Age": "<=30",   "Income": "high",   "Student": "no",  "Credit_rating": "excellent", "buys_computer": "no"},
    {"Age": "31..40", "Income": "high",   "Student": "no",  "Credit_rating": "fair",      "buys_computer": "yes"},
    {"Age": ">40",    "Income": "medium", "Student": "no",  "Credit_rating": "fair",      "buys_computer": "yes"},
    {"Age": ">40",    "Income": "low",    "Student": "yes", "Credit_rating": "fair",      "buys_computer": "yes"},
    {"Age": ">40",    "Income": "low",    "Student": "yes", "Credit_rating": "excellent", "buys_computer": "no"},
    {"Age": "31..40", "Income": "low",    "Student": "yes", "Credit_rating": "excellent", "buys_computer": "yes"},
    {"Age": "<=30",   "Income": "medium", "Student": "no",  "Credit_rating": "fair",      "buys_computer": "no"},
    {"Age": "<=30",   "Income": "low",    "Student": "yes", "Credit_rating": "fair",      "buys_computer": "yes"},
    {"Age": ">40",    "Income": "medium", "Student": "yes", "Credit_rating": "fair",      "buys_computer": "yes"},
    {"Age": "<=30",   "Income": "medium", "Student": "yes", "Credit_rating": "excellent", "buys_computer": "yes"},
    {"Age": "31..40", "Income": "medium", "Student": "no",  "Credit_rating": "excellent", "buys_computer": "yes"},
    {"Age": "31..40", "Income": "high",   "Student": "yes", "Credit_rating": "fair",      "buys_computer": "yes"},
    {"Age": ">40",    "Income": "medium", "Student": "no",  "Credit_rating": "excellent", "buys_computer": "no"}
]
```

```python
TARGET = "buys_computer"
FEATURES = ["Age", "Income", "Student", "Credit_rating"]
```

## Function to calculate entropy

```python
def entropy(rows):
    counts = Counter(row[TARGET] for row in rows)
    total = len(rows)

    result = 0

    for count in counts.values():
        probability = count / total
        result -= probability * math.log2(probability)

    return result
```

##  split data by feature Value

```python
def split_data(rows, feature):
    groups = {}

    for row in rows:
        value = row[feature]

        if value not in groups:
            groups[value] = []

        groups[value].append(row)

    return groups
```

## Calculate Expected Entropy

```python
def expected_entropy(rows, feature):
    groups = split_data(rows, feature)
    total = len(rows)
    result = 0

    for group in groups.values():
        weight = len(group) / total
        result += weight * entropy(group)

    return result
```

## Calculate Information Gain

```python
def information_gain(rows, feature):
    return entropy(rows) - expected_entropy(rows, feature)
```

## Calculate Split Information

```python
def split_information(rows, feature):
    groups = split_data(rows, feature)
    total = len(rows)
    result = 0

    for group in groups.values():
        probability = len(group) / total
        result -= probability * math.log2(probability)

    return result
```

## Calculate Gain Ratio

```python
def gain_ratio(rows, feature):
    gain = information_gain(rows, feature)
    split_info = split_information(rows, feature)

    if split_info == 0:
        return 0

    return gain / split_info
```

## Find Best feature Using C4.5

```python
print("C4.5 - GAIN RATIO CALCULATION")
print("=" * 65)

dataset_entropy = entropy(data)
print(f"Dataset Entropy: {dataset_entropy:.4f}\n")

gains = {}

print(f"{'Feature':<18}{'Expected Entropy':<20}"
      f"{'Info Gain':<14}{'Split Info':<14}{'Gain Ratio'}")
print("-" * 80)

for feature in FEATURES:
    expected = expected_entropy(data, feature)
    gain = information_gain(data, feature)
    split_info = split_information(data, feature)
    ratio = gain_ratio(data, feature)

    gains[feature] = gain

    print(f"{feature:<18}{expected:<20.4f}"
          f"{gain:<14.4f}{split_info:<14.4f}{ratio:.4f}")
```


## C4.5 safeguard:
### Consider features whose information gain is at least the average information gain.

```python
average_gain = sum(gains.values()) / len(gains)

eligible_features = [
    feature for feature in FEATURES
    if gains[feature] >= average_gain
]

best_feature = max(
    eligible_features,
    key=lambda feature: gain_ratio(data, feature)
)

print("\n" + "=" * 65)
print(f"Average Information Gain: {average_gain:.4f}")
print("Eligible features:", eligible_features)
print(f"Selected Root Feature: {best_feature}")
print(f"Highest Gain Ratio: {gain_ratio(data, best_feature):.4f}")
```
